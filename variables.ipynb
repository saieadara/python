Assigning different data types to an variables


```python
#integer variable
my_integer = 42
print(f"my_integer: (my_integer), type: {type(my_integer)}")
```

```python
#float variable
float = 3.14
print(f"float: (float), type: {type(float)}")
```

```python
#string variable
string = "Hello python"
print(f"string: (string) type: {type(string)}")
```

```python
#boolean
is_active = True
print(f"is_active: (is_active) type: {type(is_active)}")
```

```python
#reassigning a variable
price = 100
print(f"initial price: {price}")
```

```python
price =1.50
print(f"new price: {price}")
```

```python
baby_name = "stupid"
print(f"Alan walker: {baby_name}")
```

```python
#multiple assignment
x, y, z = 10, 20, 30
print(f"x:{x}, y:{y}, z:{z}")
```

```python
#swapping values
a = 5
b = 10
print(f"before swap: a:{a}, b:{b}")
a,b = b,a
print(f"after swap: a:{a}, b:{b}")
```
