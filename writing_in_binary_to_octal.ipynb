```python
# Comparing different bases of the decimal number 13
binary_val = 0b1101
octal_val = 0o15
hex_val = 0xD

# Check if they evaluate to the same decimal integer
print("Are they equal?", binary_val == octal_val == hex_val)

# Convert a hexadecimal literal directly to binary
print("Hex to Binary:", bin(0xD))
```
