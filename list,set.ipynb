```python
daily_temperature = [72,75,72,79,75,50]
print("original list:", daily_temperature)
```

```python
# Removing duplicates using set()
daily_temperatures = [72, 75, 72, 79, 75, 80]
print("Original List:", daily_temperatures)
unique_temps = set(daily_temperature)
print("As a set:", unique_temps)
# Convert back to a list for normal operations
list_items = list(unique_temps)
print("As a list:", list_items)
```

```python
my_name = ['c','c','h','e','r','r','y']
print("this is my original name:", my_name)
unique_temps = set(my_name)
print("as a set:", unique_temps)
#convert back to list in normal operations
list_items = list(unique_temps)
print("as a new list:", list_items)
```

```python
chat_gpt = [10, 20 ,30,10,20,30]
print("the original order:", chat_gpt)
sam_altman = set(chat_gpt)
print("this remove duplicates:", sam_altman)
elon_musk = list(sam_altman)
print("new items:", elon_musk)
```
